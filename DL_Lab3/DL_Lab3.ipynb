```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import time

# Task 1: Load CIFAR-10
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()
class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer', 'dog', 'frog', 'horse', 'ship', 'truck']

print(f"Training images shape: {train_images.shape}")
print(f"Training labels shape: {train_labels.shape}")
print(f"Test images shape: {test_images.shape}")
print(f"Test labels shape: {test_labels.shape}")
```

```python
# Plot 1: Sample Images
plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(train_images[i])
    plt.title(class_names[train_labels[i][0]])
    plt.axis('off')
plt.suptitle("Sample CIFAR-10 Images")
plt.savefig("sample_images.eps", format="eps", bbox_inches="tight")
plt.show()

# Plot 2: Class Distribution
unique, counts = np.unique(train_labels, return_counts=True)
plt.figure(figsize=(8, 4))
sns.barplot(x=[class_names[i] for i in unique], y=counts)
plt.title("Class Distribution")
plt.xticks(rotation=45)
plt.ylabel("Count")
plt.savefig("class_distribution.eps", format="eps", bbox_inches="tight")
plt.show()

# Task 2 & 3: Hyperparameter Experimentation
print("\n--- Task 2: Kernel Size Evaluation ---")
for k_size in [(3, 3), (5, 5), (7, 7)]:
    temp_model = models.Sequential([layers.Conv2D(32, k_size, input_shape=(32, 32, 3))])
    print(f"Kernel Size: {k_size} -> Output Shape: {temp_model.output_shape[1:]}")

print("\n--- Task 3: Stride and Padding Evaluation ---")
for stride in [1, 2]:
    for pad in ['same', 'valid']:
        temp_model = models.Sequential([layers.Conv2D(32, (3, 3), strides=stride, padding=pad, input_shape=(32, 32, 3))])
        print(f"Stride: {stride}, Padding: {pad} -> Output Shape: {temp_model.output_shape[1:]}")

# Preprocessing
X_train = train_images / 255.0
X_test = test_images / 255.0
y_train = tf.keras.utils.to_categorical(train_labels, 10)
y_test = tf.keras.utils.to_categorical(test_labels, 10)

# Task 6: CNN Architecture Construction
def create_cnn(pooling_layer=layers.MaxPooling2D, num_filters_1=32, num_filters_2=64):
    model = models.Sequential([
        layers.Conv2D(num_filters_1, (3, 3), activation='relu', input_shape=(32, 32, 3)),
        pooling_layer((2, 2)),
        layers.Conv2D(num_filters_2, (3, 3), activation='relu'),
        pooling_layer((2, 2)),
        layers.Flatten(),
        layers.Dense(64, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    return model

model_max = create_cnn(layers.MaxPooling2D)
model_max.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model_max.summary()

# Train CNN Model
history_max = model_max.fit(X_train, y_train, epochs=20, batch_size=32, validation_data=(X_test, y_test))

# Task 4: Feature Map Visualization
feature_model = tf.keras.Model(inputs=model_max.inputs, outputs=model_max.layers[0].output)
sample_img = np.expand_dims(X_test[0], axis=0)
feature_maps = feature_model.predict(sample_img)

# Plot 7: Feature Maps
plt.figure(figsize=(10, 5))
for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(feature_maps[0, :, :, i], cmap='gray')
    plt.title(f"Feature Map {i+1}")
    plt.axis('off')
plt.suptitle("8 Feature Maps from First Conv Layer")
plt.savefig("feature_maps.eps", format="eps", bbox_inches="tight")
plt.show()

# Task 6 Plots: Accuracy & Loss
# Plot 3 & 4: Training and Validation Accuracy
plt.figure(figsize=(6, 4))
plt.plot(history_max.history['accuracy'], label='Train Accuracy')
plt.plot(history_max.history['val_accuracy'], label='Val Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.savefig("accuracy_plot.eps", format="eps", bbox_inches="tight")
plt.show()

# Plot 5 & 6: Training and Validation Loss
plt.figure(figsize=(6, 4))
plt.plot(history_max.history['loss'], label='Train Loss')
plt.plot(history_max.history['val_loss'], label='Val Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.savefig("loss_plot.eps", format="eps", bbox_inches="tight")
plt.show()


# Task 5: Max Pooling vs Average Pooling
model_avg = create_cnn(layers.AveragePooling2D)
model_avg.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
history_avg = model_avg.fit(X_train, y_train, epochs=20, batch_size=32, validation_data=(X_test, y_test), verbose=0)

loss_max, acc_max = model_max.evaluate(X_test, y_test, verbose=0)
loss_avg, acc_avg = model_avg.evaluate(X_test, y_test, verbose=0)

print(f"\nMax Pooling Test Accuracy: {acc_max:.4f}")
print(f"Average Pooling Test Accuracy: {acc_avg:.4f}")

# Task 7: Performance Evaluation
y_pred_prob = model_max.predict(X_test)
y_pred = np.argmax(y_pred_prob, axis=1)
y_true = test_labels.flatten()

print("\nClassification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

# Plot 8: Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("True")
plt.savefig("confusion_matrix.eps", format="eps", bbox_inches="tight")
plt.show()



```
